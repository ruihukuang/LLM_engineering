````python
import os
from twilio.rest import Client
````

````python
    # Your Account SID and Auth Token from twilio.com/console
    # It's recommended to set these as environment variables for security
    account_sid = os.getenv("TWILIO_ACCOUNT_SID")
    auth_token = os.getenv("TWILIO_AUTH_TOKEN")
    me_from = os.getenv('TWILIO_FROM', 'your-phone-number-if-not-using-env')
    me_to = os.getenv('MY_PHONE_NUMBER', 'your-phone-number-if-not-using-env')

    client = Client(account_sid, auth_token)

    message = client.messages.create(
        body="Hello from Twilio!",
        from_=me_from,  # Your Twilio phone number
        to=me_to     # The recipient's phone number
    )

    print(f"Message SID: {message.sid}")
    print(f"Message Status: {message.status}")
````
